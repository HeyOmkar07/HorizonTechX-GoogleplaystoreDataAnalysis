```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("googleplaystore.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df['Installs'] = df['Installs'].str.replace('+','')
df['Installs'] = df['Installs'].str.replace(',','')

df['Installs'] = pd.to_numeric(df['Installs'], errors='coerce')
```

```python
df['Category'].value_counts().head(10)
```

```python
df['Category'].value_counts().head(10).plot(kind='bar')
plt.title("Top App Categories")
plt.show()
```

```python
df.info()
```

```python
df.head()
```

```python
category_installs = df.groupby('Category')['Installs'].sum().sort_values(ascending=False).head(10)

category_installs.plot(kind='bar')
plt.title('Top Categories by Installs')
plt.xlabel('Category')
plt.ylabel('Total Installs')
plt.show()
```

```python
df.to_csv("clean_googleplaystore.csv", index=False)
```
